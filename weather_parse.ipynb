```python
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None) 
```

```python
df_weather = pd.DataFrame(columns=['station_index', 
                                   'station_name', 
                                   'temperature_avg',
                                   'temperature_normal',
                                   'temperature_dev',
                                   'precipitation_avg',
                                   'precipitation_normal',
                                   'precipitation_dev_persent',
                                   'region',
                                   'month',
                                   'year'
                                  ])
region=''
list_year=[2022,2023,2024,2025]
for lit_year in list_year:
      for lit_month in range(1,12):
              tables=pd.read_html(f"http://pogodaiklimat.ru/monitors.php?id=rus&month={lit_month}&year={lit_year}",header=0)
              df=tables[0]
              for index,row in df.iterrows():
                          if row['Индекс станции']==row['Название']:
                                  region=row['Название']
                          if row['Индекс станции']!=row['Название'] and row['Индекс станции']!='Индекс станции':
                                  new_stat= {'station_index':str(row['Индекс станции']), 
                                             'station_name':str(row['Название']), 
                                             'temperature_avg':str(row['Температура воздуха, °С']),
                                             'temperature_normal':str(row['Температура воздуха, °С.1']),
                                             'temperature_dev':str(row['Температура воздуха, °С.2']),
                                             'precipitation_avg':str(row['Осадки, мм']),
                                             'precipitation_normal':str(row['Осадки, мм.1']),
                                             'precipitation_dev_persent':str(row['Осадки, мм.2']),
                                             'region':region,
                                             'month':str(lit_month),
                                             'year':str(lit_year)}
                                  df_weather.loc[len(df_weather)] = new_stat         
df_weather.head(20)
```

```python
df_weather.to_csv('weather_parse.csv')
```
